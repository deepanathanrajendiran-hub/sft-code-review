# Code Reviewer SFT + Dr. GRPO

**Run order:**
1. Cell 1-2: Install deps
2. Cell 3: Upload `train_dataset_clean.jsonl` to Colab
3. Cell 4-5: Rewrite training data with Gemini (~$1.39, ~30-60 min)
4. Cell 6-9: SFT training + test
5. Cell 10-11: Save & merge SFT model
6. Cell 12: Load merged SFT model for GRPO
7. Cell 13: Prepare GRPO dataset + define reward function
8. Cell 14: Dr. GRPO training
9. Cell 15: Test GRPO model
10. Cell 16: Save & download

**Why Dr. GRPO over DPO?**
- No preference pairs needed — eliminates the weak link (SFT-generated "rejected" samples were too similar to "chosen")
- No reference model — simpler setup, less VRAM
- Rule-based reward function — you define exactly what "good code review" means
- Dr. GRPO removes per-question std normalization bias and length bias from vanilla GRPO

````python
# ===== CELL 4: Rewrite Training Data with Gemini =====
# Cost: ~$1.39 with Gemini Flash
# Time: ~30-60 min for 12k samples
# Supports resume — safe to re-run if interrupted

import os
import vertexai
from vertexai.generative_models import GenerativeModel, HarmCategory, HarmBlockThreshold
from tqdm import tqdm
import json
import time

PROJECT_ID = "halogen-sol-348516"
LOCATION = "us-central1"
INPUT_FILE = "train_dataset_clean.jsonl"
OUTPUT_FILE = "train_dataset_rewritten.jsonl"

vertexai.init(project=PROJECT_ID, location=LOCATION)

active_model = None
for name in ["gemini-2.0-flash-001", "gemini-1.5-flash"]:
    try:
        m = GenerativeModel(name)
        m.generate_content("test")
        active_model = m
        print(f"Connected to: {name}")
        break
    except:
        continue

if not active_model:
    raise RuntimeError("Could not connect to any Gemini model.")

# Load input data
data = []
with open(INPUT_FILE, 'r') as f:
    for line in f:
        data.append(json.loads(line))
print(f"Loaded {len(data)} samples")

# Resume logic
processed_hashes = set()
if os.path.exists(OUTPUT_FILE):
    with open(OUTPUT_FILE, 'r') as f:
        for line in f:
            try:
                entry = json.loads(line)
                processed_hashes.add(hash(entry.get('input', '')[:500]))
            except:
                pass
    print(f"Resuming: {len(processed_hashes)} already done")
````

````python
# ===== CELL 5: Run Rewriting Loop (PARALLEL — 20 workers) =====
# ~15-30 min instead of 5 hrs
from concurrent.futures import ThreadPoolExecutor, as_completed
import threading

safety = {HarmCategory.HARM_CATEGORY_DANGEROUS_CONTENT: HarmBlockThreshold.BLOCK_ONLY_HIGH}

REWRITE_PROMPT = """You are a Senior Software Engineer writing a code review for a pull request.

Given a code diff and an existing review comment, rewrite the review to be:
1. SPECIFIC - Reference actual code, variable names, or line changes from the diff
2. ACTIONABLE - Tell the developer exactly what to change and why
3. CRITICAL - If there's a bug, security issue, or bad practice, call it out clearly
4. CONCISE - 1-3 sentences max

If the diff has a security issue (hardcoded secrets, SQL injection, etc.), start with "SECURITY:" and explain the risk.
If the diff has a bug (missing null checks, wrong logic, etc.), start with "BUG:" and explain what will break.
If the diff is just a style/naming issue, explain why the naming matters.

If the diff is trivial (whitespace, formatting only) and the review adds no value, output: {{"keep": false}}

Diff:
{diff}

Existing review:
{comment}

Output JSON only: {{"keep": true/false, "rewritten": "your improved review"}}"""

stats = {"total": 0, "rewritten": 0, "filtered": 0, "skipped": 0, "errors": 0}
file_lock = threading.Lock()
stats_lock = threading.Lock()

# Filter out already-processed samples
samples_to_process = []
for sample in data:
    diff_hash = hash(sample['input'][:500])
    if diff_hash in processed_hashes:
        with stats_lock:
            stats["skipped"] += 1
    else:
        samples_to_process.append(sample)

print(f"Skipped {stats['skipped']} already processed. Processing {len(samples_to_process)} remaining...")

output_file = open(OUTPUT_FILE, "a")

def process_sample(sample):
    diff_hash = hash(sample['input'][:500])

    for attempt in range(5):
        try:
            response = active_model.generate_content(
                REWRITE_PROMPT.format(
                    diff=sample['input'][:5000],
                    comment=sample['output'],
                ),
                generation_config={"response_mime_type": "application/json"},
                safety_settings=safety,
            )

            try:
                raw = response.text.replace("```json", "").replace("```", "").strip()
                result = json.loads(raw)
            except json.JSONDecodeError:
                with stats_lock:
                    stats["errors"] += 1
                return

            keep_val = result.get("keep")
            if keep_val in [True, "true", "True"]:
                rewritten = result.get("rewritten", "").strip()
                if rewritten and len(rewritten) >= 20:
                    entry = {
                        "instruction": "Review this code change.",
                        "input": sample['input'],
                        "output": rewritten,
                        "repo": sample.get('repo', ''),
                    }
                    with file_lock:
                        output_file.write(json.dumps(entry) + "\n")
                        output_file.flush()
                        processed_hashes.add(diff_hash)
                    with stats_lock:
                        stats["rewritten"] += 1
            else:
                with stats_lock:
                    stats["filtered"] += 1

            return  # Success

        except Exception as e:
            if "429" in str(e) or "ResourceExhausted" in str(e):
                time.sleep(20 * (attempt + 1))
                continue
            else:
                with stats_lock:
                    stats["errors"] += 1
                    if stats["errors"] <= 5:
                        print(f"\nError: {str(e)[:100]}")
                return

# Run parallel
MAX_WORKERS = 20
with ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:
    futures = {executor.submit(process_sample, s): s for s in samples_to_process}
    for future in tqdm(as_completed(futures), total=len(futures), unit="sample"):
        future.result()
        with stats_lock:
            stats["total"] += 1
        if stats["total"] % 1000 == 0:
            print(f"\nProgress: {stats}")

output_file.close()
print(f"\nDone! {stats}")
print(f"Output: {OUTPUT_FILE}")
````

````python
# ===== CELL 6: Load Model for SFT =====
from unsloth import FastLanguageModel
import torch

max_seq_length = 8192

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="unsloth/Qwen2.5-Coder-7B-Instruct",
    max_seq_length=max_seq_length,
    dtype=None,
    load_in_4bit=False,
)

# FIX: pad_token != eos_token (Qwen2.5 bug)
tokenizer.pad_token = "<|vision_pad|>"

model = FastLanguageModel.get_peft_model(
    model,
    r=32,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj",
                    "gate_proj", "up_proj", "down_proj"],
    lora_alpha=32,
    lora_dropout=0,
    bias="none",
    use_gradient_checkpointing="unsloth",
    random_state=42,
)

print("Model loaded")
````

````python
# ===== CELL 7: Load & Format Rewritten Dataset =====
import json
from datasets import Dataset

data = []
with open('train_dataset_rewritten.jsonl', 'r') as f:
    for line in f:
        data.append(json.loads(line))

print(f"Loaded {len(data)} samples")

SYSTEM_MSG = "You are a Senior Software Engineer reviewing code changes. Provide clear, actionable feedback."

def format_prompt(sample):
    """Use tokenizer.apply_chat_template — NO manual ChatML."""
    messages = [
        {"role": "system", "content": SYSTEM_MSG},
        {"role": "user", "content": f"Review this code diff:\n\n```diff\n{sample['input'][:6000]}\n```"},
        {"role": "assistant", "content": sample['output']},
    ]
    text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=False)
    return {"text": text}

dataset = Dataset.from_list(data)
dataset = dataset.map(format_prompt)
dataset = dataset.train_test_split(test_size=0.05, seed=42)
print(f"Train: {len(dataset['train'])}, Val: {len(dataset['test'])}")

# Verify: check the formatted text looks correct
print("\n--- Sample (first 500 chars) ---")
print(dataset['train'][0]['text'][:500])
````

````python
# ===== CELL 8: SFT Training =====
from trl import SFTTrainer
from transformers import TrainingArguments

trainer = SFTTrainer(
    model=model,
    tokenizer=tokenizer,
    train_dataset=dataset["train"],
    eval_dataset=dataset["test"],
    dataset_text_field="text",
    max_seq_length=max_seq_length,
    dataset_num_proc=4,
    packing=True,
    args=TrainingArguments(
        per_device_train_batch_size=4,
        gradient_accumulation_steps=4,
        warmup_steps=50,
        num_train_epochs=3,
        learning_rate=5e-5,
        fp16=False,
        bf16=True,
        logging_steps=25,
        optim="adamw_8bit",
        weight_decay=0.01,
        lr_scheduler_type="cosine",
        seed=42,
        output_dir="outputs-sft-v2",
        eval_strategy="steps",
        eval_steps=100,
        save_strategy="steps",
        save_steps=200,
    ),
)

print("Starting SFT training...")
trainer_stats = trainer.train()
print("SFT training complete")
````

````python
# ===== CELL 9: Test SFT Model =====
FastLanguageModel.for_inference(model)

test_diffs = [
    ("SECURITY", """@@ -12,7 +12,7 @@ def connect(self):\n-        password = os.environ.get('DB_PASS')\n+        password = \"admin123\"\n         return db.connect(password)"""),
    ("BUG", """@@ -5,7 +5,7 @@ def get_user(user_id):\n-        user = db.query(User).filter(User.id == user_id).first()\n+        user = db.query(User).filter(User.id == user_id)\n         return user.name"""),
    ("STYLE", """@@ -1,5 +1,5 @@\n-def calculateTotalPrice(items):\n+def calc(i):\n     total = 0\n-    for item in items:\n+    for x in i:\n         total += x.price"""),
]

for i, (label, diff) in enumerate(test_diffs, 1):
    messages = [
        {"role": "system", "content": SYSTEM_MSG},
        {"role": "user", "content": f"Review this code diff:\n\n```diff\n{diff}\n```"},
    ]
    inputs = tokenizer.apply_chat_template(
        messages, tokenize=True, add_generation_prompt=True, return_tensors="pt"
    ).to("cuda")

    with torch.no_grad():
        outputs = model.generate(
            input_ids=inputs, max_new_tokens=256, temperature=0.3,
            top_p=0.9, pad_token_id=tokenizer.pad_token_id,
        )

    response = tokenizer.decode(outputs[0], skip_special_tokens=True)
    review = response.split("assistant")[-1].strip()
    print(f"\n{'='*50}\nTEST {i} ({label})\n{'='*50}\nMODEL REVIEW: {review}")
````

````python
# ===== CELL 10: Save SFT LoRA =====
model.save_pretrained("code-reviewer-lora-v2")
tokenizer.save_pretrained("code-reviewer-lora-v2")
print("SFT LoRA saved")
````

````python
# ===== CELL 11: Merge SFT into Base (needed for DPO reference) =====
model.save_pretrained_merged("sft-merged-v2", tokenizer, save_method="merged_16bit")
print("SFT model merged into base")
````

````python
# ===== CELL 12: Load Merged SFT + LoRA for Dr. GRPO =====
# No reference model needed — Dr. GRPO uses group-relative advantages
import gc
import torch

# Free SFT model memory
try:
    del model
    del trainer
except:
    pass
torch.cuda.empty_cache()
gc.collect()
print("GPU memory freed")

from unsloth import FastLanguageModel

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="sft-merged-v2",
    max_seq_length=4096,
    dtype=None,
    load_in_4bit=False,
)

tokenizer.pad_token = "<|vision_pad|>"

# LoRA for GRPO — smaller rank, conservative updates
model = FastLanguageModel.get_peft_model(
    model,
    r=16,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj",
                    "gate_proj", "up_proj", "down_proj"],
    lora_alpha=16,
    lora_dropout=0,
    bias="none",
    use_gradient_checkpointing="unsloth",
    random_state=42,
)

SYSTEM_MSG = "You are a Senior Software Engineer reviewing code changes. Provide clear, actionable feedback."
print("Merged SFT model loaded with LoRA for Dr. GRPO")
````

````python
# ===== CELL 13: Prepare GRPO Dataset + Reward Function =====
# GRPO only needs prompts — no preference pairs!
# The model generates its own completions and learns from reward signals.
import json
import re
from datasets import Dataset

data = []
with open('train_dataset_rewritten.jsonl', 'r') as f:
    for line in f:
        data.append(json.loads(line))
print(f"Loaded {len(data)} samples")

# Build dataset: prompt (chat messages) + diff_text (raw diff for reward fn)
grpo_data = []
for sample in data:
    diff = sample['input'][:4000]
    grpo_data.append({
        "prompt": [
            {"role": "system", "content": SYSTEM_MSG},
            {"role": "user", "content": f"Review this code diff:\n\n```diff\n{diff}\n```"},
        ],
        "diff_text": diff,  # Passed to reward function as kwarg
    })

dataset = Dataset.from_list(grpo_data)
dataset = dataset.train_test_split(test_size=0.05, seed=42)
print(f"Train: {len(dataset['train'])}, Eval: {len(dataset['test'])}")

# --- REWARD FUNCTION ---
# Multi-signal, rule-based — no learned reward model needed.
# Each signal adds/subtracts from the score. The model learns to maximize total reward.

def code_review_reward(completions, diff_text, **kwargs):
    """
    Reward signals:
    1. Category prefix (SECURITY:/BUG:) when the diff warrants it  (+3.0)
    2. Specificity — backtick-quoted code references               (+2.0/+1.0)
    3. Line number references                                      (+1.0)
    4. Actionability — contains action verbs                        (+1.0)
    5. Conciseness — 1-3 sentences                                  (+1.0)
    6. Good length — 50-400 chars                                   (+1.0)
    7. Penalize vague/generic reviews                               (-2.0)
    8. Penalize empty/too-short reviews                             (-3.0)
    """
    rewards = []
    for review, diff in zip(completions, diff_text):
        review = review.strip()
        score = 0.0

        # Empty/too-short → hard penalty
        if len(review) < 20:
            rewards.append(-3.0)
            continue

        # 1. Good length (50-400 chars)
        if 50 <= len(review) <= 400:
            score += 1.0
        elif len(review) > 600:
            score -= 1.0

        # 2. Category prefix
        has_prefix = any(review.startswith(p) for p in ["SECURITY:", "BUG:"])
        if has_prefix:
            score += 1.0

        # 3. Specificity — backtick-quoted code references
        code_refs = re.findall(r'`[^`]+`', review)
        if len(code_refs) >= 1:
            score += 2.0
        if len(code_refs) >= 2:
            score += 1.0

        # 4. Line number references
        if re.search(r'[Ll]ine\s*\d+|L\d+', review):
            score += 1.0

        # 5. Actionability — contains action verbs
        actions = ['rename', 'change', 'use', 'add', 'remove', 'replace',
                   'revert', 'fix', 'update', 'move', 'extract', 'wrap',
                   'avoid', 'ensure', 'validate', 'check', 'instead']
        if any(w in review.lower() for w in actions):
            score += 1.0

        # 6. Penalize vague/generic reviews
        vague = ['looks good', 'lgtm', 'please fix', 'needs work',
                 'could be improved', 'not great', 'nice change']
        if any(v in review.lower() for v in vague):
            score -= 2.0

        # 7. Conciseness (1-3 sentences ideal)
        sentences = [s.strip() for s in re.split(r'[.!?]+', review) if s.strip()]
        if 1 <= len(sentences) <= 3:
            score += 1.0
        elif len(sentences) > 5:
            score -= 1.0

        # 8. Security detection — big bonus for correct prefix
        security_kw = ['password', 'secret', 'api_key', 'token', 'credential',
                       'hardcod', 'sql injection', 'eval(', 'exec(']
        has_security = any(k in diff.lower() for k in security_kw)
        if has_security and review.startswith("SECURITY:"):
            score += 3.0
        elif has_security and not review.startswith("SECURITY:"):
            score -= 1.0

        # 9. Bug detection — bonus for catching removed safety checks
        bug_kw = ['.first()', '.get(', 'is None', 'is not None',
                  'try:', 'except:', '!= null', '!== null']
        has_bug = any(k in diff for k in bug_kw)
        if has_bug and review.startswith("BUG:"):
            score += 3.0

        rewards.append(score)
    return rewards

# Sanity check
test_reviews = [
    "SECURITY: The `password` on line 14 is hardcoded to `admin123`. Use `os.environ.get('DB_PASS')` instead.",
    "looks good to me",
    "rename the function",
]
test_diffs = [
    'password = "admin123"',
    "x = 1",
    "def calc(i):",
]
print(f"\nReward sanity check: {code_review_reward(test_reviews, test_diffs)}")
# Expected: ~[12.0, -2.0, 1.0]
````

````python
# ===== CELL 14: Dr. GRPO Training =====
# Dr. GRPO = GRPO with debiased advantages:
#   - Removes per-question std normalization (only mean-centering)
#   - Removes response-length normalization
# This prevents over-weighting easy questions and rewarding long wrong answers.

from trl import GRPOTrainer, GRPOConfig

trainer = GRPOTrainer(
    model=model,
    processing_class=tokenizer,
    reward_funcs=[code_review_reward],
    args=GRPOConfig(
        loss_type="dr_grpo",
        num_generations=8,            # 8 completions per prompt for advantage estimation
        max_completion_length=256,     # Reviews should be concise
        max_prompt_length=1500,
        per_device_train_batch_size=2,
        gradient_accumulation_steps=4, # Effective batch = 2 * 4 = 8 prompts (64 completions)
        num_train_epochs=1,            # 1 epoch is typically enough for GRPO
        learning_rate=5e-6,
        beta=0.04,                     # KL penalty — keeps model close to SFT
        warmup_ratio=0.1,
        bf16=True,
        logging_steps=10,
        optim="adamw_8bit",
        weight_decay=0.01,
        seed=42,
        output_dir="outputs-grpo-v1",
        eval_strategy="steps",
        eval_steps=50,
        save_strategy="steps",
        save_steps=200,
    ),
    train_dataset=dataset["train"],
    eval_dataset=dataset["test"],
)

print("Starting Dr. GRPO training...")
trainer.train()
print("Dr. GRPO training complete")
````

````python
# ===== CELL 15: Test Dr. GRPO Model =====
FastLanguageModel.for_inference(model)

test_diffs = [
    ("SECURITY", """@@ -12,7 +12,7 @@ def connect(self):\n-        password = os.environ.get('DB_PASS')\n+        password = \"admin123\"\n         return db.connect(password)"""),
    ("BUG", """@@ -5,7 +5,7 @@ def get_user(user_id):\n-        user = db.query(User).filter(User.id == user_id).first()\n+        user = db.query(User).filter(User.id == user_id)\n         return user.name"""),
    ("STYLE", """@@ -1,5 +1,5 @@\n-def calculateTotalPrice(items):\n+def calc(i):\n     total = 0\n-    for item in items:\n+    for x in i:\n         total += x.price"""),
]

for i, (label, diff) in enumerate(test_diffs, 1):
    messages = [
        {"role": "system", "content": SYSTEM_MSG},
        {"role": "user", "content": f"Review this code diff:\n\n```diff\n{diff}\n```"},
    ]
    inputs = tokenizer.apply_chat_template(
        messages, tokenize=True, add_generation_prompt=True, return_tensors="pt"
    ).to("cuda")

    with torch.no_grad():
        outputs = model.generate(
            input_ids=inputs, max_new_tokens=256, temperature=0.3,
            top_p=0.9, pad_token_id=tokenizer.pad_token_id,
        )

    response = tokenizer.decode(outputs[0], skip_special_tokens=True)
    review = response.split("assistant")[-1].strip()

    # Also show the reward score
    reward = code_review_reward([review], [diff])[0]

    print(f"\n{'='*50}")
    print(f"TEST {i} ({label}) — Reward: {reward}")
    print(f"{'='*50}")
    print(f"DIFF:\n{diff}")
    print(f"\nMODEL REVIEW:\n{review}")
````
