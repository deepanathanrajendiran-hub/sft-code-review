# Dr. GRPO for Code Review

**Loads pre-trained SFT (or DPO) adapter, applies Dr. GRPO reinforcement learning, and evaluates Base vs SFT/DPO vs Dr. GRPO with quantitative metrics.**

**Stack: transformers + PEFT + TRL + Liger Kernel + SDPA + vLLM (eval + GRPO generation)**

**Run order:**
1. Cells 1-3: Install deps, upload/locate files + unzip SFT/DPO weights, init W&B
2. Cells 4-5: Load dataset + reward fn, define eval metrics (BLEU/ROUGE/reward/win-rate)
3. Cell 6: Merge SFT or DPO adapter into base — set `UPSTREAM_MODEL = "sft"` or `"dpo"`
4. Cell 7: vLLM subprocess eval — BASE and SFT/DPO (~30-60 sec each on RunPod)
5. Cells 8-9: Load merged model + LoRA, Dr. GRPO training (Liger Kernel + vLLM generation, W&B logged)
6. Cell 10: Merge GRPO + vLLM subprocess eval
7. Cells 11-13: Comparison table + charts, win-rate matrix + side-by-side, save

**Metrics tracked:**
- BLEU-4, ROUGE-1/2/L (vs Gemini reference reviews)
- Mean reward score (from rule-based reward fn)
- Pairwise win rate (Base vs SFT/DPO vs Dr. GRPO)
- Side-by-side qualitative examples

**Required files:** `code-reviewer-lora-v2.zip` (SFT) or `code-reviewer-dpo-v3.zip` (DPO), `train_dataset_rewritten.jsonl`

````python
# ===== CELL 4: Load Dataset + Define Reward Function (Haiku Judge) =====
import json, re, random
import anthropic
from concurrent.futures import ThreadPoolExecutor, as_completed
from rouge_score import rouge_scorer as rouge_scorer_lib

SYSTEM_MSG = "You are a Senior Software Engineer reviewing code changes. Provide clear, actionable feedback."

# Load rewritten data
_data_path = DATA_FILE if 'DATA_FILE' in dir() else 'train_dataset_rewritten.jsonl'
data = []
with open(_data_path, 'r') as f:
    for line in f:
        data.append(json.loads(line))
print(f"Loaded {len(data)} samples from {_data_path}")

random.seed(42)
indices = list(range(len(data)))
random.shuffle(indices)

EVAL_SIZE = 200
eval_indices = set(indices[:EVAL_SIZE])
eval_data = [data[i] for i in sorted(eval_indices)]
train_data = [data[i] for i in sorted(set(indices) - eval_indices)]
print(f"Train: {len(train_data)}, Eval: {len(eval_data)}")

# --- Haiku Judge Setup ---
_haiku_client = anthropic.Anthropic()
_rl_scorer = rouge_scorer_lib.RougeScorer(['rougeL'], use_stemmer=True)  # fallback
_haiku_errors = 0

def _judge_single(review, diff, reference):
    """Call Haiku to score a single code review. Returns 0-10."""
    global _haiku_errors
    try:
        resp = _haiku_client.messages.create(
            model="claude-haiku-4-5-20251001",
            max_tokens=4,
            messages=[{
                "role": "user",
                "content": (
                    "Rate this code review from 0 to 10.\n\n"
                    f"CODE DIFF:\n{diff[:2000]}\n\n"
                    f"REFERENCE (high-quality example):\n{reference[:500]}\n\n"
                    f"REVIEW TO EVALUATE:\n{review[:500]}\n\n"
                    "Criteria: accuracy (does it find real issues?), "
                    "actionability (are fixes specific?), "
                    "relevance (focuses on important parts of diff?).\n"
                    "Score anchors:\n"
                    "  5 = average — correct but generic, could apply to any diff.\n"
                    "  7+ = requires specific line references, variable names, or function names from the diff.\n"
                    "  3 or below = vague advice that could be written without reading the diff.\n"
                    "Return ONLY a single integer 0-10."
                ),
            }],
        )
        text = resp.content[0].text.strip()
        match = re.search(r'\d+', text)
        if match:
            return min(max(int(match.group()), 0), 10)
        return 5
    except Exception as e:
        _haiku_errors += 1
        if _haiku_errors <= 3:
            print(f"[WARN] Haiku API error ({_haiku_errors}): {e}")
        # Fallback to ROUGE-L
        if reference:
            rl = _rl_scorer.score(reference, review)['rougeL'].fmeasure
            return int(rl * 10)
        return 5

# --- Helper functions ---
def _extract_text(completion):
    if isinstance(completion, str):
        return completion
    if isinstance(completion, list):
        for msg in completion:
            if isinstance(msg, dict) and msg.get("role") == "assistant":
                return msg.get("content", "")
        return " ".join(m.get("content", "") if isinstance(m, dict) else str(m) for m in completion)
    return str(completion)

def _repeat_to_match(lst, target_len):
    n = len(lst)
    if n == target_len:
        return lst
    if target_len % n == 0:
        return [x for x in lst for _ in range(target_len // n)]
    return (lst * ((target_len // n) + 1))[:target_len]

_debug_count = 0

def code_review_reward(completions, **kwargs):
    global _debug_count
    _debug_count += 1

    diff_text = kwargs.get("diff_text", None)
    reference_review = kwargs.get("reference_review", None)

    n = len(completions)
    diff_text = _repeat_to_match(diff_text or [""] * n, n)
    reference_review = _repeat_to_match(reference_review or [""] * n, n)

    # Extract all review texts
    reviews = [_extract_text(c).strip() for c in completions]

    if _debug_count <= 2:
        print(f"\n[DEBUG call {_debug_count}] n={n}, reviews[0][:150]: {reviews[0][:150]}")

    # --- Phase 1: Batch Haiku judge calls in parallel ---
    haiku_scores = [5] * n
    with ThreadPoolExecutor(max_workers=16) as executor:
        futures = {}
        for i, (review, diff, ref) in enumerate(zip(reviews, diff_text, reference_review)):
            if len(review) >= 20:
                futures[executor.submit(_judge_single, review, diff, ref)] = i
        for future in as_completed(futures):
            idx = futures[future]
            try:
                haiku_scores[idx] = future.result()
            except Exception:
                haiku_scores[idx] = 5

    if _debug_count <= 2:
        print(f"[DEBUG call {_debug_count}] haiku_scores: {haiku_scores[:8]}")

    # --- Phase 2: Combine Haiku + structural signals ---
    security_kw = ['password', 'secret', 'api_key', 'token', 'credential',
                   'hardcod', 'sql injection', 'eval(', 'exec(']
    bug_kw_ci = ['is none', 'is not none', '!= null', '!== null']
    bug_kw_cs = ['.first()', '.get(', 'try:', 'except:']
    vague_kw = ['looks good', 'lgtm', 'please fix', 'needs work', 'could be improved']

    rewards = []
    for i, (review, diff, ref) in enumerate(zip(reviews, diff_text, reference_review)):
        if len(review) < 20:
            rewards.append(-3.0)
            continue

        # Primary: Haiku judge (0-6 pts)
        score = haiku_scores[i] / 10.0 * 6.0

        # Secondary: structural checks (reduced weights — Haiku handles quality)
        diff_lower = diff.lower()
        has_security = any(k in diff_lower for k in security_kw)
        has_bug = any(k in diff_lower for k in bug_kw_ci) or any(k in diff for k in bug_kw_cs)

        if has_security:
            score += 1.5 if review.startswith("SECURITY:") else -1.0
        elif review.startswith("SECURITY:"):
            score -= 1.0
        if has_bug and review.startswith("BUG:"):
            score += 1.0
        elif review.startswith("BUG:") and not has_bug:
            score -= 0.5

        # Code refs in diff (min 3 chars)
        code_refs = re.findall(r'`([^`]{3,})`', review)
        matched = sum(1 for c in code_refs if c in diff)
        score += min(matched, 2) * 0.5

        # Vague phrases
        if any(v in review.lower() for v in vague_kw):
            score -= 1.0

        score = max(min(score, 8.0), -3.0)
        rewards.append(score)

    assert len(rewards) == n
    return rewards

# --- Sanity check (live Haiku call) ---
test_r = code_review_reward(
    ["SECURITY: The `password` on line 14 is hardcoded. Use `os.environ.get('DB_PASS')`.",
     "looks good to me",
     "SECURITY: This looks fine to me"],
    diff_text=['password = "admin123"', "x = 1", "x = 1"],
    reference_review=[
        "SECURITY: Hardcoded `password` on line 14. Replace with `os.environ.get('DB_PASS')`.",
        "No issues found.",
        "No issues found."
    ]
)
print(f"Reward sanity check: {test_r}")
assert all(-3.0 <= r <= 8.0 for r in test_r), f"Rewards out of bounds: {test_r}"
assert test_r[0] > test_r[1], f"Good review should score higher than vague one"
print(f"Haiku judge working — good review: {test_r[0]:.1f}, vague: {test_r[1]:.1f}, false prefix: {test_r[2]:.1f}")
print(f"Haiku API errors so far: {_haiku_errors}")
_debug_count = 0
````

````python
# ===== CELL 5: Define Eval Metrics (BLEU / ROUGE / Reward / Win-Rate) =====
# Uses vLLM in an isolated subprocess — ~30-60 sec per model on RunPod
import os, json, subprocess
import numpy as np
from rouge_score import rouge_scorer
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction

_scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)
smooth = SmoothingFunction().method1

def compute_metrics(predictions, references, diffs):
    bleu_scores, rouge1_scores, rouge2_scores, rougeL_scores = [], [], [], []
    for pred, ref in zip(predictions, references):
        bleu = sentence_bleu([ref.split()], pred.split(), smoothing_function=smooth)
        bleu_scores.append(bleu)
        r = _scorer.score(ref, pred)
        rouge1_scores.append(r['rouge1'].fmeasure)
        rouge2_scores.append(r['rouge2'].fmeasure)
        rougeL_scores.append(r['rougeL'].fmeasure)
    rewards = code_review_reward(predictions, diff_text=diffs)
    return {
        "bleu4":       np.mean(bleu_scores),
        "rouge1":      np.mean(rouge1_scores),
        "rouge2":      np.mean(rouge2_scores),
        "rougeL":      np.mean(rougeL_scores),
        "mean_reward": np.mean(rewards),
        "reward_std":  np.std(rewards),
    }

def generate_reviews_vllm(model_path, eval_samples, tokenizer_name=None):
    """Run vLLM eval in an isolated subprocess. ~30-60 sec per model for 200 samples on RunPod."""
    if tokenizer_name is None:
        tokenizer_name = model_path

    samples_path = "/tmp/_eval_samples.json"
    output_path  = "/tmp/_eval_output.json"
    script_path  = "/tmp/_eval_script.py"

    with open(samples_path, "w") as f:
        json.dump(eval_samples, f)

    SYSTEM_MSG = "You are a Senior Software Engineer reviewing code changes. Provide clear, actionable feedback."

    script_lines = [
        "import json, gc",
        "from vllm import LLM, SamplingParams",
        "from transformers import AutoTokenizer",
        "",
        f"with open('{samples_path}') as f:",
        "    eval_samples = json.load(f)",
        "",
        f"SYSTEM_MSG = {repr(SYSTEM_MSG)}",
        f"tok = AutoTokenizer.from_pretrained({repr(tokenizer_name)})",
        "",
        "prompts, references, diffs = [], [], []",
        "for sample in eval_samples:",
        "    diff = sample['input'][:3000]",  # matches training cap
        "    messages = [",
        "        {'role': 'system', 'content': SYSTEM_MSG},",
        "        {'role': 'user', 'content': 'Review this code diff:\\\\n\\\\n```diff\\\\n' + diff + '\\\\n```'},",
        "    ]",
        "    text = tok.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)",
        "    prompts.append(text)",
        "    references.append(sample['output'])",
        "    diffs.append(diff)",
        "",
        "llm = LLM(",
        f"    model={repr(model_path)},",
        f"    tokenizer={repr(tokenizer_name)},",
        "    max_model_len=4096,",
        "    dtype='bfloat16',",
        "    gpu_memory_utilization=0.80,",
        "    tensor_parallel_size=1,",
        ")",
        "sampling_params = SamplingParams(temperature=0, max_tokens=256)",
        "outputs = llm.generate(prompts, sampling_params)",
        "predictions = [o.outputs[0].text.strip() for o in outputs]",
        "",
        "del llm",
        "gc.collect()",
        "",
        f"with open('{output_path}', 'w') as f:",
        "    json.dump({'predictions': predictions, 'references': references, 'diffs': diffs}, f)",
        "print(f'EVAL_DONE — {len(predictions)} predictions')",
    ]

    with open(script_path, "w") as f:
        f.write("\n".join(script_lines))

    print(f"  Evaluating {model_path} (vLLM greedy)...")
    result = subprocess.run(
        ["python", script_path],
        capture_output=True, text=True, timeout=600,
    )

    if result.returncode != 0:
        print("STDOUT:", result.stdout[-1000:])
        print("STDERR:", result.stderr[-2000:])
        raise RuntimeError(f"Eval subprocess failed (exit code {result.returncode})")

    with open(output_path) as f:
        data = json.load(f)

    for p in [samples_path, output_path, script_path]:
        try:
            os.unlink(p)
        except OSError:
            pass

    return data["predictions"], data["references"], data["diffs"]

def pairwise_win_rate(preds_a, preds_b, diffs, margin=0.5):
    rewards_a = code_review_reward(preds_a, diff_text=diffs)
    rewards_b = code_review_reward(preds_b, diff_text=diffs)
    a_wins = sum(1 for ra, rb in zip(rewards_a, rewards_b) if ra > rb + margin)
    b_wins = sum(1 for ra, rb in zip(rewards_a, rewards_b) if rb > ra + margin)
    ties   = len(rewards_a) - a_wins - b_wins
    n      = len(rewards_a)
    return a_wins / n, b_wins / n, ties / n

def pairwise_from_rewards(rewards_a, rewards_b, margin=0.5):
    a_wins = sum(1 for ra, rb in zip(rewards_a, rewards_b) if ra > rb + margin)
    b_wins = sum(1 for ra, rb in zip(rewards_a, rewards_b) if rb > ra + margin)
    ties   = len(rewards_a) - a_wins - b_wins
    n      = len(rewards_a)
    return a_wins / n, b_wins / n, ties / n

print("Eval helpers defined (vLLM subprocess, greedy, gpu_util=0.80)")
````

````python
# ===== CELL 6: Merge DPO Adapter into Base =====
UPSTREAM_MODEL = "dpo"  # "sft" → SFT→GRPO | "dpo" → SFT→DPO→GRPO

if UPSTREAM_MODEL == "dpo":
    LORA_PATH   = "code-reviewer-dpo-v4"
    MERGED_NAME = "dpo-merged-for-grpo"
    print("Pipeline: SFT → DPO → GRPO")
else:
    LORA_PATH   = "code-reviewer-lora-v2"
    MERGED_NAME = "sft-merged-for-grpo"
    print("Pipeline: SFT → GRPO")

from peft import PeftModel
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch, gc

BASE_MODEL = "unsloth/Qwen2.5-Coder-7B-Instruct"

print(f"Loading base model on CPU (keeps GPU free for vLLM eval)...")
base_model = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL,
    torch_dtype=torch.bfloat16,
    device_map="cpu",
)
tokenizer = AutoTokenizer.from_pretrained(BASE_MODEL)

print(f"Merging {UPSTREAM_MODEL.upper()} LoRA from {LORA_PATH}/ ...")
peft_model = PeftModel.from_pretrained(base_model, LORA_PATH, device_map="cpu")
merged = peft_model.merge_and_unload()

print(f"Saving merged model to {MERGED_NAME}/ ...")
merged.save_pretrained(MERGED_NAME)
tokenizer.save_pretrained(MERGED_NAME)

del merged, peft_model, base_model, tokenizer
gc.collect()
print(f"Done — {MERGED_NAME}/ ready (GPU untouched)")
````

````python
# ===== CELL 7: Batch Eval — BASE and SFT/DPO =====
# vLLM subprocess — ~30-60 sec per model on RunPod

# --- Eval BASE model ---
print("Evaluating BASE model...")
base_preds, base_refs, base_diffs = generate_reviews_vllm(
    model_path="unsloth/Qwen2.5-Coder-7B-Instruct",
    eval_samples=eval_data,
)
base_metrics = compute_metrics(base_preds, base_refs, base_diffs)

print(f"\n--- BASE Model Metrics ---")
for k, v in base_metrics.items():
    print(f"  {k}: {v:.4f}")
wandb.log({f"base/{k}": v for k, v in base_metrics.items()})

# --- Eval SFT/DPO model ---
upstream_label = UPSTREAM_MODEL.upper()
print(f"\nEvaluating {upstream_label} model ({MERGED_NAME})...")
sft_preds, sft_refs, sft_diffs = generate_reviews_vllm(
    model_path=MERGED_NAME,
    eval_samples=eval_data,
)
sft_metrics = compute_metrics(sft_preds, sft_refs, sft_diffs)

print(f"\n--- {upstream_label} Model Metrics ---")
for k, v in sft_metrics.items():
    print(f"  {k}: {v:.4f}")
wandb.log({f"{UPSTREAM_MODEL}/{k}": v for k, v in sft_metrics.items()})

# Show 3 sample comparisons
for i in range(3):
    print(f"\n{'='*60}")
    print(f"DIFF (first 200 chars): {base_diffs[i][:200]}")
    print(f"BASE REVIEW:          {base_preds[i]}")
    print(f"{upstream_label} REVIEW: {sft_preds[i]}")
    print(f"REFERENCE:            {base_refs[i]}")
````

````python
# ===== CELL 8: Load Merged SFT/DPO + LoRA for Dr. GRPO (transformers + PEFT) =====
from transformers import AutoModelForCausalLM, AutoTokenizer
from peft import LoraConfig, get_peft_model
import torch

# SDPA: built into PyTorch 2.x — uses flash kernels on A100/L4 automatically, no install needed
model = AutoModelForCausalLM.from_pretrained(
    MERGED_NAME,
    torch_dtype=torch.bfloat16,
    attn_implementation="sdpa",
    device_map="auto",
)
# use_reentrant=False required for Liger Kernel compatibility
model.gradient_checkpointing_enable(gradient_checkpointing_kwargs={"use_reentrant": False})

tokenizer = AutoTokenizer.from_pretrained(MERGED_NAME)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
    model.config.pad_token_id = tokenizer.pad_token_id

model = get_peft_model(model, LoraConfig(
    r=16,
    lora_alpha=32,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj",
                    "gate_proj", "up_proj", "down_proj"],
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM",
))
model.print_trainable_parameters()
print(f"{UPSTREAM_MODEL.upper()} model loaded with LoRA for Dr. GRPO (SDPA attention)")
````

````python
# ===== CELL 9: Dr. GRPO Training (Liger Kernel + SDPA + vLLM generation) =====
from datasets import Dataset
from trl import GRPOTrainer, GRPOConfig
import trl, inspect
print(f"TRL version: {trl.__version__}")

# Build GRPO dataset — cap diff at 3000 chars to fit max_prompt_length=1500 tokens
# Use all available training data (13k total - 200 eval = ~12800 samples)
grpo_data = []
for sample in train_data:
    diff = sample['input'][:3000]
    grpo_data.append({
        "prompt": [
            {"role": "system", "content": SYSTEM_MSG},
            {"role": "user", "content": f"Review this code diff:\n\n```diff\n{diff}\n```"},
        ],
        "diff_text": diff,
        "reference_review": sample['output'][:500],
    })

grpo_dataset = Dataset.from_list(grpo_data)
print(f"GRPO Train: {len(grpo_dataset)} samples")

valid_params = set(inspect.signature(GRPOConfig.__init__).parameters.keys())

desired = dict(
    use_liger_kernel=True,
    use_vllm=True,
    vllm_gpu_memory_utilization=0.5,
    vllm_enable_sleep_mode=True,
    vllm_server_host="0.0.0.0",
    vllm_server_port=8000,
    loss_type="dr_grpo",
    num_generations=8,
    generation_batch_size=8,
    max_completion_length=256,
    max_prompt_length=1500,
    per_device_train_batch_size=2,
    gradient_accumulation_steps=4,
    num_train_epochs=1,
    learning_rate=5e-6,
    beta=0.04,
    warmup_ratio=0.1,
    bf16=True,
    logging_steps=10,
    optim="adamw_8bit",
    weight_decay=0.01,
    seed=42,
    output_dir="outputs-grpo-v1",
    eval_strategy="no",
    save_strategy="steps",
    save_steps=100,
    report_to="wandb",
    dataloader_num_workers=2,
    dataloader_pin_memory=True,
)

filtered = {k: v for k, v in desired.items() if k in valid_params}
skipped = set(desired) - set(filtered)
if skipped:
    print(f"Skipped unsupported params: {skipped}")
critical = {"loss_type", "learning_rate", "beta", "num_generations"}
dropped_critical = critical & skipped
if dropped_critical:
    raise RuntimeError(f"Critical params not supported by TRL {trl.__version__}: {dropped_critical}")
print(f"Using params: {sorted(filtered.keys())}")

trainer = GRPOTrainer(
    model=model,
    processing_class=tokenizer,
    reward_funcs=[code_review_reward],
    args=GRPOConfig(**filtered),
    train_dataset=grpo_dataset,
)

print("Starting Dr. GRPO training (Liger Kernel + SDPA + vLLM)...")
trainer.train()
print("Dr. GRPO training complete")
````

````python
# ===== CELL 10: Merge GRPO + vLLM Batch Eval =====
import gc
import torch

# Merge GRPO LoRA into the SFT/DPO base (PEFT merge)
merged_model = model.merge_and_unload()
merged_model.save_pretrained("grpo-merged-for-eval")
tokenizer.save_pretrained("grpo-merged-for-eval")
print("GRPO model merged and saved to grpo-merged-for-eval/")

# Free training model — subprocess gets full GPU
del merged_model, model, trainer
gc.collect()
torch.cuda.empty_cache()

# vLLM eval (subprocess — gets full GPU)
print("\nEvaluating Dr. GRPO model...")
grpo_preds, grpo_refs, grpo_diffs = generate_reviews_vllm(
    model_path="grpo-merged-for-eval",
    eval_samples=eval_data,
)
grpo_metrics = compute_metrics(grpo_preds, grpo_refs, grpo_diffs)

print(f"\n--- Dr. GRPO Model Metrics ---")
for k, v in grpo_metrics.items():
    print(f"  {k}: {v:.4f}")
wandb.log({f"grpo/{k}": v for k, v in grpo_metrics.items()})

for i in range(3):
    print(f"\n{'='*60}")
    print(f"DIFF (first 200 chars): {grpo_diffs[i][:200]}")
    print(f"GRPO REVIEW: {grpo_preds[i]}")
    print(f"REFERENCE:   {grpo_refs[i]}")
````

````python
# ===== CELL 11: Comparison Table + Bar Charts =====
import matplotlib.pyplot as plt

upstream_label = UPSTREAM_MODEL.upper()

# Build comparison dict
all_metrics = {
    "Base": base_metrics,
    upstream_label: sft_metrics,
    "Dr. GRPO": grpo_metrics,
}

# --- Print Table ---
metric_names = ["bleu4", "rouge1", "rouge2", "rougeL", "mean_reward", "reward_std"]
header = f"{'Metric':<15} {'Base':>10} {upstream_label:>10} {'Dr. GRPO':>10} {upstream_label+' D':>10} {'GRPO D':>10}"
print("=" * len(header))
print(header)
print("=" * len(header))
for m in metric_names:
    base_v = base_metrics[m]
    sft_v = sft_metrics[m]
    grpo_v = grpo_metrics[m]
    sft_delta = sft_v - base_v
    grpo_delta = grpo_v - base_v
    sign_s = "+" if sft_delta >= 0 else ""
    sign_g = "+" if grpo_delta >= 0 else ""
    print(f"{m:<15} {base_v:>10.4f} {sft_v:>10.4f} {grpo_v:>10.4f} {sign_s}{sft_delta:>9.4f} {sign_g}{grpo_delta:>9.4f}")
print("=" * len(header))

# Log table to W&B
table = wandb.Table(columns=["Metric", "Base", upstream_label, "Dr. GRPO"])
for m in metric_names:
    table.add_data(m, base_metrics[m], sft_metrics[m], grpo_metrics[m])
wandb.log({"comparison_table": table})

# --- Bar Charts ---
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
colors = ["#4C72B0", "#55A868", "#C44E52"]
models = ["Base", upstream_label, "Dr. GRPO"]

# Chart 1: BLEU & ROUGE
text_metrics = ["bleu4", "rouge1", "rouge2", "rougeL"]
x = np.arange(len(text_metrics))
width = 0.25
for i, model_name in enumerate(models):
    vals = [all_metrics[model_name][m] for m in text_metrics]
    axes[0].bar(x + i * width, vals, width, label=model_name, color=colors[i])
axes[0].set_ylabel("Score")
axes[0].set_title("BLEU / ROUGE Scores")
axes[0].set_xticks(x + width)
axes[0].set_xticklabels(text_metrics)
axes[0].legend()
axes[0].set_ylim(0, 1)

# Chart 2: Mean Reward
reward_vals = [all_metrics[m]["mean_reward"] for m in models]
reward_stds = [all_metrics[m]["reward_std"] for m in models]
axes[1].bar(models, reward_vals, color=colors, yerr=reward_stds, capsize=5)
axes[1].set_ylabel("Mean Reward")
axes[1].set_title("Mean Reward Score")

# Chart 3: Improvement over Base
improve_metrics = ["bleu4", "rouge1", "rougeL", "mean_reward"]
x2 = np.arange(len(improve_metrics))
for i, model_name in enumerate([upstream_label, "Dr. GRPO"]):
    deltas = [all_metrics[model_name][m] - all_metrics["Base"][m] for m in improve_metrics]
    axes[2].bar(x2 + i * 0.35, deltas, 0.35, label=model_name, color=colors[i + 1])
axes[2].set_ylabel("D vs Base")
axes[2].set_title("Improvement Over Base")
axes[2].set_xticks(x2 + 0.175)
axes[2].set_xticklabels(improve_metrics)
axes[2].legend()
axes[2].axhline(y=0, color='black', linewidth=0.5)

plt.tight_layout()
plt.savefig("eval_comparison.png", dpi=150, bbox_inches='tight')
wandb.log({"eval_charts": wandb.Image("eval_comparison.png")})
plt.show()
print("Charts saved to eval_comparison.png")
````
