# Code Review Model Evaluation — Full Pipeline
**Base → SFT → DPO → GRPO**

Independent evaluation using **Claude Haiku as pairwise judge** — no circular reward function.

**Metrics:**
- BLEU-4, ROUGE-1/2/L (surface similarity to reference reviews)
- Haiku pairwise win-rates (semantic quality judged by independent LLM)
- Side-by-side qualitative examples

**Cost:** ~$0.50 for all Haiku judge calls | **Time:** ~10-15 min total

````python
# ===== CELL 2: Setup + Load Eval Data =====
import os, json, random, zipfile

os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"
assert os.environ.get("ANTHROPIC_API_KEY"), "Set ANTHROPIC_API_KEY for Haiku judge"

# --- Model paths (adjust if your dirs differ) ---
BASE_MODEL = "unsloth/Qwen2.5-Coder-7B-Instruct"
SFT_ZIP    = os.environ.get("SFT_ZIP_PATH", "/workspace/code-reviewer-lora-v2.zip")
SFT_LORA   = "code-reviewer-lora-v2"
SFT_MERGED = "sft-merged-for-eval"
DPO_MERGED = "dpo-merged-for-grpo"       # created by dr.grpo.ipynb Cell 7
GRPO_MERGED = "grpo-merged-for-eval"     # created by dr.grpo.ipynb Cell 10
DATA_FILE  = os.environ.get("DATA_PATH", "/workspace/train_dataset_rewritten.jsonl")

# --- Load eval dataset (same split as training) ---
data = []
with open(DATA_FILE) as f:
    for line in f:
        data.append(json.loads(line))

random.seed(42)
indices = list(range(len(data)))
random.shuffle(indices)
EVAL_SIZE = 200
eval_data = [data[i] for i in sorted(indices[:EVAL_SIZE])]
print(f"Loaded {len(data)} total, {len(eval_data)} eval samples")
````

````python
# ===== CELL 3: Merge SFT Adapter into Base =====
import torch, gc

if os.path.isdir(SFT_MERGED):
    print(f"{SFT_MERGED}/ already exists — skipping merge")
else:
    # Unzip if needed
    if not os.path.isdir(SFT_LORA) and os.path.exists(SFT_ZIP):
        print(f"Extracting {SFT_ZIP}...")
        with zipfile.ZipFile(SFT_ZIP, "r") as zf:
            zf.extractall(".")

    assert os.path.isdir(SFT_LORA), f"SFT adapter not found: {SFT_LORA}/"

    from transformers import AutoModelForCausalLM, AutoTokenizer
    from peft import PeftModel

    print("Loading base model on CPU...")
    base = AutoModelForCausalLM.from_pretrained(BASE_MODEL, torch_dtype=torch.bfloat16, device_map="cpu")
    tok = AutoTokenizer.from_pretrained(BASE_MODEL)

    print(f"Merging SFT LoRA from {SFT_LORA}/...")
    peft_model = PeftModel.from_pretrained(base, SFT_LORA, device_map="cpu")
    merged = peft_model.merge_and_unload()
    merged.save_pretrained(SFT_MERGED)
    tok.save_pretrained(SFT_MERGED)

    del merged, peft_model, base, tok
    gc.collect()
    print(f"Saved to {SFT_MERGED}/")

# Verify all model dirs exist
for name, path in [("DPO", DPO_MERGED), ("GRPO", GRPO_MERGED), ("SFT", SFT_MERGED)]:
    assert os.path.isdir(path), f"{name} model not found: {path}/"
    print(f"{name}: {path}/ OK")
````

````python
# ===== CELL 4: Generate Reviews for All 4 Models (vLLM subprocess) =====
import subprocess, json, os

SYSTEM_MSG = "You are a Senior Software Engineer reviewing code changes. Provide clear, actionable feedback."

def generate_reviews_vllm(model_path, eval_samples, tokenizer_name=None):
    if tokenizer_name is None:
        tokenizer_name = model_path

    samples_path = "/tmp/_eval_samples.json"
    output_path  = "/tmp/_eval_output.json"
    script_path  = "/tmp/_eval_script.py"

    with open(samples_path, "w") as f:
        json.dump(eval_samples, f)

    script = f"""
import json, gc
from vllm import LLM, SamplingParams
from transformers import AutoTokenizer

with open('{samples_path}') as f:
    eval_samples = json.load(f)

SYSTEM_MSG = {repr(SYSTEM_MSG)}
tok = AutoTokenizer.from_pretrained({repr(tokenizer_name)})

prompts, references, diffs = [], [], []
for sample in eval_samples:
    diff = sample['input'][:3000]
    messages = [
        {{'role': 'system', 'content': SYSTEM_MSG}},
        {{'role': 'user', 'content': 'Review this code diff:\\n\\n```diff\\n' + diff + '\\n```'}},
    ]
    text = tok.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    prompts.append(text)
    references.append(sample['output'])
    diffs.append(diff)

llm = LLM(
    model={repr(model_path)},
    tokenizer={repr(tokenizer_name)},
    max_model_len=4096,
    dtype='bfloat16',
    gpu_memory_utilization=0.85,
    tensor_parallel_size=1,
)
outputs = llm.generate(prompts, SamplingParams(temperature=0, max_tokens=256))
predictions = [o.outputs[0].text.strip() for o in outputs]
del llm
gc.collect()

with open('{output_path}', 'w') as f:
    json.dump({{'predictions': predictions, 'references': references, 'diffs': diffs}}, f)
print(f'EVAL_DONE - {{len(predictions)}} predictions')
"""
    with open(script_path, "w") as f:
        f.write(script)

    print(f"  Evaluating {model_path}...")
    result = subprocess.run(["python", script_path], capture_output=True, text=True, timeout=600)
    if result.returncode != 0:
        print("STDERR:", result.stderr[-2000:])
        raise RuntimeError(f"Eval failed for {model_path}")

    with open(output_path) as f:
        out = json.load(f)
    for p in [samples_path, output_path, script_path]:
        try: os.unlink(p)
        except: pass
    return out["predictions"], out["references"], out["diffs"]

# --- Generate for all 4 models ---
print("=" * 60)
print("Generating reviews for all 4 models (vLLM greedy, 200 samples each)")
print("=" * 60)

print("\n[1/4] BASE")
base_preds, base_refs, base_diffs = generate_reviews_vllm(BASE_MODEL, eval_data)

print("\n[2/4] SFT")
sft_preds, sft_refs, sft_diffs = generate_reviews_vllm(SFT_MERGED, eval_data)

print("\n[3/4] DPO")
dpo_preds, dpo_refs, dpo_diffs = generate_reviews_vllm(DPO_MERGED, eval_data)

print("\n[4/4] GRPO")
grpo_preds, grpo_refs, grpo_diffs = generate_reviews_vllm(GRPO_MERGED, eval_data)

print("\nAll 4 models evaluated!")
````

````python
# ===== CELL 5: Standard Metrics (BLEU / ROUGE / Pass@1) =====
import numpy as np, re
from rouge_score import rouge_scorer
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
from concurrent.futures import ThreadPoolExecutor, as_completed
import anthropic

_scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)
_smooth = SmoothingFunction().method1
_score_client = anthropic.Anthropic()
_score_errors = 0

PASS_THRESHOLD        = 7   # "good review"
PASS_THRESHOLD_STRICT = 8   # "excellent review"

def haiku_score_review(review, diff, reference):
    """Rate a single review 0-10 via Haiku. Returns int."""
    global _score_errors
    try:
        resp = _score_client.messages.create(
            model="claude-haiku-4-5-20251001",
            max_tokens=4,
            messages=[{"role": "user", "content": (
                "Rate this code review 0 to 10.\n\n"
                f"CODE DIFF:\n{diff[:2000]}\n\n"
                f"REFERENCE (expert review):\n{reference[:500]}\n\n"
                f"REVIEW TO RATE:\n{review[:500]}\n\n"
                "Criteria: accuracy, actionability, specificity, relevance.\n"
                "Reply ONLY with a single integer 0-10."
            )}],
        )
        text = resp.content[0].text.strip()
        m = re.search(r'\d+', text)
        return min(max(int(m.group()), 0), 10) if m else 5
    except Exception as e:
        _score_errors += 1
        if _score_errors <= 3:
            print(f"[WARN] Score error ({_score_errors}): {e}")
        return 5

def compute_haiku_scores(predictions, diffs, refs):
    """Score all predictions in parallel. Returns list of 0-10 ints."""
    scores = [5] * len(predictions)
    with ThreadPoolExecutor(max_workers=16) as executor:
        futures = {executor.submit(haiku_score_review, p, d, r): i
                   for i, (p, d, r) in enumerate(zip(predictions, diffs, refs))}
        for future in as_completed(futures):
            scores[futures[future]] = future.result()
    return scores

def compute_metrics(predictions, references):
    bleu, r1, r2, rl = [], [], [], []
    for pred, ref in zip(predictions, references):
        bleu.append(sentence_bleu([ref.split()], pred.split(), smoothing_function=_smooth))
        r = _scorer.score(ref, pred)
        r1.append(r['rouge1'].fmeasure)
        r2.append(r['rouge2'].fmeasure)
        rl.append(r['rougeL'].fmeasure)
    return {
        "bleu4": np.mean(bleu), "rouge1": np.mean(r1),
        "rouge2": np.mean(r2), "rougeL": np.mean(rl),
    }

base_metrics = compute_metrics(base_preds, base_refs)
sft_metrics  = compute_metrics(sft_preds,  sft_refs)
dpo_metrics  = compute_metrics(dpo_preds,  dpo_refs)
grpo_metrics = compute_metrics(grpo_preds, grpo_refs)

# --- Pass@1 via Haiku individual scoring (800 calls, ~$0.10) ---
print("Computing Pass@1 scores (Haiku individual scoring, 4 models x 200 samples)...")
base_scores = compute_haiku_scores(base_preds, base_diffs, base_refs)
sft_scores  = compute_haiku_scores(sft_preds,  base_diffs, base_refs)
dpo_scores  = compute_haiku_scores(dpo_preds,  base_diffs, base_refs)
grpo_scores = compute_haiku_scores(grpo_preds, base_diffs, base_refs)
print(f"Done. Haiku scoring errors: {_score_errors}")

for scores, metrics in [
    (base_scores, base_metrics),
    (sft_scores,  sft_metrics),
    (dpo_scores,  dpo_metrics),
    (grpo_scores, grpo_metrics),
]:
    metrics["haiku_mean"]       = float(np.mean(scores))
    metrics["pass_at_1"]        = sum(s >= PASS_THRESHOLD        for s in scores) / len(scores)
    metrics["pass_at_1_strict"] = sum(s >= PASS_THRESHOLD_STRICT for s in scores) / len(scores)

print("\nMetrics (BLEU / ROUGE / Pass@1):")
header = f"{'Metric':<22} {'Base':>8} {'SFT':>8} {'DPO':>8} {'GRPO':>8}"
print("=" * len(header))
print(header)
print("=" * len(header))
for m in ["bleu4", "rouge1", "rouge2", "rougeL", "haiku_mean", "pass_at_1", "pass_at_1_strict"]:
    print(f"{m:<22} {base_metrics[m]:>8.4f} {sft_metrics[m]:>8.4f} {dpo_metrics[m]:>8.4f} {grpo_metrics[m]:>8.4f}")
print("=" * len(header))
````

````python
# ===== CELL 6: Haiku Pairwise Judge =====
import anthropic, re, random
from concurrent.futures import ThreadPoolExecutor, as_completed

_client = anthropic.Anthropic()
_judge_errors = 0

def haiku_pairwise_judge(review_a, review_b, diff, reference):
    """Ask Haiku which review is better. Returns 'A', 'B', or 'TIE'."""
    global _judge_errors

    # Random swap to eliminate position bias
    swap = random.random() > 0.5
    ra, rb = (review_b, review_a) if swap else (review_a, review_b)

    try:
        resp = _client.messages.create(
            model="claude-haiku-4-5-20251001",
            max_tokens=4,
            messages=[{"role": "user", "content": (
                "Compare two code reviews for the same diff. Which is better?\n\n"
                f"DIFF:\n{diff[:2000]}\n\n"
                f"REFERENCE (expert review):\n{reference[:500]}\n\n"
                f"REVIEW A:\n{ra[:500]}\n\n"
                f"REVIEW B:\n{rb[:500]}\n\n"
                "Criteria: accuracy, actionability, specificity, relevance.\n"
                "Reply ONLY: A, B, or TIE"
            )}],
        )
        result = resp.content[0].text.strip().upper()
        # Normalize
        if "TIE" in result:
            result = "TIE"
        elif "A" in result and "B" not in result:
            result = "A"
        elif "B" in result and "A" not in result:
            result = "B"
        else:
            result = "TIE"

        # Undo swap
        if swap:
            if result == "A": result = "B"
            elif result == "B": result = "A"
        return result
    except Exception as e:
        _judge_errors += 1
        if _judge_errors <= 5:
            print(f"[WARN] Judge error ({_judge_errors}): {e}")
        return "TIE"

def run_pairwise_eval(preds_a, preds_b, diffs, refs, label=""):
    """Run pairwise Haiku judge on 200 samples in parallel. Returns (a_win%, b_win%, tie%)."""
    results = [None] * len(preds_a)

    with ThreadPoolExecutor(max_workers=16) as executor:
        futures = {}
        for i in range(len(preds_a)):
            f = executor.submit(haiku_pairwise_judge, preds_a[i], preds_b[i], diffs[i], refs[i])
            futures[f] = i
        for future in as_completed(futures):
            results[futures[future]] = future.result()

    a_wins = sum(1 for r in results if r == "A")
    b_wins = sum(1 for r in results if r == "B")
    ties   = sum(1 for r in results if r == "TIE")
    n = len(results)

    if label:
        print(f"  {label}: A={a_wins/n:.1%}  B={b_wins/n:.1%}  Tie={ties/n:.1%}")
    return a_wins / n, b_wins / n, ties / n, results

# --- Run all 6 pairwise comparisons ---
print("Running Haiku pairwise evaluation (6 comparisons x 200 samples)...")
print("This takes ~2-3 min and costs ~$0.50\n")

wr = {}  # win_results[(model_a, model_b)] = (a_win%, b_win%, tie%)

_, _, _, _ = run_pairwise_eval(sft_preds, base_preds, base_diffs, base_refs, "SFT vs Base")
wr[("SFT", "Base")] = _[:3]

_, _, _, _ = run_pairwise_eval(dpo_preds, base_preds, base_diffs, base_refs, "DPO vs Base")
wr[("DPO", "Base")] = _[:3]

_, _, _, grpo_vs_base_results = run_pairwise_eval(grpo_preds, base_preds, base_diffs, base_refs, "GRPO vs Base")
wr[("GRPO", "Base")] = (grpo_vs_base_results.count("A")/200, grpo_vs_base_results.count("B")/200, grpo_vs_base_results.count("TIE")/200)

_, _, _, _ = run_pairwise_eval(dpo_preds, sft_preds, base_diffs, base_refs, "DPO vs SFT")
wr[("DPO", "SFT")] = _[:3]

_, _, _, _ = run_pairwise_eval(grpo_preds, sft_preds, base_diffs, base_refs, "GRPO vs SFT")
wr[("GRPO", "SFT")] = _[:3]

_, _, _, grpo_vs_dpo_results = run_pairwise_eval(grpo_preds, dpo_preds, base_diffs, base_refs, "GRPO vs DPO")
wr[("GRPO", "DPO")] = (grpo_vs_dpo_results.count("A")/200, grpo_vs_dpo_results.count("B")/200, grpo_vs_dpo_results.count("TIE")/200)

print(f"\nHaiku judge errors: {_judge_errors}")
print("Done!")
````

````python
# ===== CELL 8: Win-Rate Heatmap =====
import matplotlib.pyplot as plt
import numpy as np

models = ["Base", "SFT", "DPO", "GRPO"]
n = len(models)
matrix = np.full((n, n), 0.5)  # diagonal = 0.5

# Fill matrix from wr dict
model_idx = {m: i for i, m in enumerate(models)}
for (a, b), (aw, bw, tw) in wr.items():
    i, j = model_idx[a], model_idx[b]
    matrix[i][j] = aw       # A's win rate against B
    matrix[j][i] = bw       # B's win rate against A

fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(matrix, cmap="RdYlGn", vmin=0, vmax=1)

ax.set_xticks(range(n))
ax.set_yticks(range(n))
ax.set_xticklabels(models, fontsize=12)
ax.set_yticklabels(models, fontsize=12)
ax.set_xlabel("Model B", fontsize=12)
ax.set_ylabel("Model A", fontsize=12)
ax.set_title("Pairwise Win Rate Matrix (Haiku Judge)\nRow model A win rate vs Column model B", fontsize=13)

for i in range(n):
    for j in range(n):
        val = matrix[i][j]
        color = "white" if val > 0.65 or val < 0.35 else "black"
        ax.text(j, i, f"{val:.1%}", ha="center", va="center", color=color, fontsize=13, fontweight="bold")

plt.colorbar(im, ax=ax, label="Win Rate")
plt.tight_layout()
plt.savefig("win_rate_heatmap.png", dpi=150, bbox_inches="tight")
plt.show()
print("Heatmap saved to win_rate_heatmap.png")
````
