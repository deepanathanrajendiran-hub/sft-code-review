````python
# ===== CELL 3: Merge SFT LoRA into Base (CPU — keeps GPU free) =====
# Only need to run this once — skips if sft-merged-for-dpo/ already exists
import os, gc, torch
from peft import PeftModel
from transformers import AutoModelForCausalLM, AutoTokenizer

MERGED_DIR = "sft-merged-for-dpo"

if os.path.isdir(MERGED_DIR):
    print(f"{MERGED_DIR}/ already exists — skipping merge")
else:
    BASE = "unsloth/Qwen2.5-Coder-7B-Instruct"
    print("Loading base model on CPU (~2 min)...")
    base = AutoModelForCausalLM.from_pretrained(BASE, torch_dtype=torch.bfloat16, device_map="cpu")
    tok = AutoTokenizer.from_pretrained(BASE)

    print("Merging SFT LoRA...")
    peft_model = PeftModel.from_pretrained(base, "code-reviewer-lora-v2", device_map="cpu")
    merged = peft_model.merge_and_unload()
    merged.save_pretrained(MERGED_DIR)
    tok.save_pretrained(MERGED_DIR)

    del merged, peft_model, base, tok
    gc.collect()
    print(f"Saved to {MERGED_DIR}/")
````

````python
# ===== CELL 4: Load DPO Dataset =====
import json
from datasets import Dataset

SYSTEM_MSG = "You are a Senior Software Engineer reviewing code changes. Provide clear, actionable feedback."

dpo_data = []
with open('dpo_pairs.jsonl', 'r') as f:
    for line in f:
        dpo_data.append(json.loads(line))

print(f"Loaded {len(dpo_data)} DPO pairs")
print(f"Keys: {list(dpo_data[0].keys())}")
print(f"\nSample chosen:   {dpo_data[0]['chosen'][0]['content'][:120]}")
print(f"Sample rejected: {dpo_data[0]['rejected'][0]['content'][:120]}")

dataset = Dataset.from_list(dpo_data).train_test_split(test_size=0.05, seed=42)
print(f"\nTrain: {len(dataset['train'])} | Val: {len(dataset['test'])}")
````

````python
# ===== CELL 5: Init Weights & Biases =====
import wandb

wandb.login()
wandb.init(
    project="code-reviewer-dpo",
    name="dpo-v4-fixed",
    config={
        "base_model": "unsloth/Qwen2.5-Coder-7B-Instruct",
        "beta": 0.05,
        "learning_rate": 2e-6,
        "num_train_epochs": 1,
        "max_steps": 500,
        "early_stopping_patience": 3,
        "lora_r": 16,
        "max_length": 2048,
    },
)
print("W&B initialized")
````

````python
# ===== CELL 6: Load Merged SFT + New LoRA for DPO =====
from transformers import AutoModelForCausalLM, AutoTokenizer
from peft import LoraConfig, get_peft_model
import torch

model = AutoModelForCausalLM.from_pretrained(
    "sft-merged-for-dpo",
    torch_dtype=torch.bfloat16,
    attn_implementation="sdpa",   # PyTorch native — uses flash kernels on A100
    device_map="auto",
)
model.gradient_checkpointing_enable(gradient_checkpointing_kwargs={"use_reentrant": False})

tokenizer = AutoTokenizer.from_pretrained("sft-merged-for-dpo")
tokenizer.pad_token = "<|vision_pad|>"
tokenizer.padding_side = "left"
model.config.pad_token_id = tokenizer.convert_tokens_to_ids("<|vision_pad|>")

lora_config = LoraConfig(
    r=16,
    lora_alpha=16,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj",
                    "gate_proj", "up_proj", "down_proj"],
    lora_dropout=0,
    bias="none",
    task_type="CAUSAL_LM",
)
model = get_peft_model(model, lora_config)
model.print_trainable_parameters()
print("Model loaded with DPO LoRA (SDPA, no Unsloth needed on A100)")
````

````python
# ===== CELL 7: DPO Training =====
from trl import DPOTrainer, DPOConfig
from transformers import EarlyStoppingCallback

# No PatchDPOTrainer — plain TRL supports ref_model=None (LoRA disable trick) natively
dpo_trainer = DPOTrainer(
    model=model,
    ref_model=None,       # ref = base model via LoRA disable trick (no extra memory)
    args=DPOConfig(
        # ── Core DPO settings (anti-reward-hacking) ──
        beta=0.05,
        loss_type="sigmoid",

        # ── Optimizer ──
        learning_rate=2e-6,
        per_device_train_batch_size=2,
        gradient_accumulation_steps=4,
        optim="adamw_8bit",
        warmup_ratio=0.1,

        # ── Stop early — val loss plateaued at step ~400 in v3 ──
        num_train_epochs=1,
        max_steps=500,
        bf16=True,
        seed=42,

        # ── Eval + early stopping ──
        eval_strategy="steps",
        eval_steps=50,
        save_strategy="steps",
        save_steps=100,
        load_best_model_at_end=True,
        metric_for_best_model="eval_loss",
        greater_is_better=False,

        # ── Sequence lengths ──
        max_length=2048,
        max_prompt_length=1500,

        logging_steps=10,
        remove_unused_columns=False,
        output_dir="outputs-dpo-v4",
        report_to="wandb",
    ),
    train_dataset=dataset["train"],
    eval_dataset=dataset["test"],
    processing_class=tokenizer,
    callbacks=[EarlyStoppingCallback(early_stopping_patience=3)],
)

print("Starting DPO training (beta=0.05, lr=2e-6, early stopping)...")
print("Watch W&B: reward margin should stay < 2.0")
dpo_trainer.train()
print("DPO training complete")
````

````python
# ===== CELL 9: Systematic Val Eval — ROUGE + Reward =====
# Compare DPO vs SFT reference. If DPO is worse → don't use for GRPO.
import json
import numpy as np
import torch
from rouge_score import rouge_scorer as rs_module

eval_scorer = rs_module.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)

# Load val data (last 100 samples from rewritten dataset)
# Upload train_dataset_rewritten.jsonl if available, otherwise skip this cell
import os
if not os.path.exists('train_dataset_rewritten.jsonl'):
    print("train_dataset_rewritten.jsonl not found — skipping systematic eval")
    print("Upload it to compare DPO vs SFT on held-out data")
else:
    all_data = []
    with open('train_dataset_rewritten.jsonl') as f:
        for line in f:
            all_data.append(json.loads(line))
    val_data = all_data[-100:]

    SYSTEM_MSG = "You are a Senior Software Engineer reviewing code changes. Provide clear, actionable feedback."

    def quick_eval(mdl, tok, samples, n=50):
        preds, refs = [], []
        for s in samples[:n]:
            diff = s['input'][:6000]
            msgs = [
                {"role": "system", "content": SYSTEM_MSG},
                {"role": "user",   "content": f"Review this code diff:\n\n```diff\n{diff}\n```"},
            ]
            inp = tok.apply_chat_template(msgs, tokenize=True, add_generation_prompt=True, return_tensors="pt").to("cuda")
            with torch.no_grad():
                out = mdl.generate(input_ids=inp, max_new_tokens=256, do_sample=False, pad_token_id=tok.pad_token_id)
            text = tok.decode(out[0], skip_special_tokens=True).split("assistant")[-1].strip()
            preds.append(text)
            refs.append(s['output'])
        r1, r2, rl = [], [], []
        for p, r in zip(preds, refs):
            sc = eval_scorer.score(r, p)
            r1.append(sc['rouge1'].fmeasure)
            r2.append(sc['rouge2'].fmeasure)
            rl.append(sc['rougeL'].fmeasure)
        return {"rouge1": np.mean(r1), "rouge2": np.mean(r2), "rougeL": np.mean(rl)}, preds

    print("Running DPO model on 50 val samples...")
    dpo_metrics, dpo_preds = quick_eval(model, tokenizer, val_data)

    print("\nDPO v4 val metrics:")
    for k, v in dpo_metrics.items():
        print(f"  {k}: {v:.4f}")

    wandb.log({f"dpo_val/{k}": v for k, v in dpo_metrics.items()})

    print("\n--- Decision guide ---")
    print("If ROUGE-L >= 0.25 and reviews have SECURITY:/BUG: prefixes:")
    print("  → set UPSTREAM_MODEL='dpo' in dr.grpo.ipynb Cell 6")
    print("If ROUGE-L < 0.20 or reviews are generic:")
    print("  → keep UPSTREAM_MODEL='sft' (DPO still not helping)")
````
