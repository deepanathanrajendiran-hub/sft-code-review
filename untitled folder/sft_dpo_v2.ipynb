````python
# ===== CELL 4: Rewrite Training Data with Gemini =====
# Cost: ~$1.39 with Gemini Flash
# Time: ~30-60 min for 12k samples
# Supports resume — safe to re-run if interrupted

import os
import vertexai
from vertexai.generative_models import GenerativeModel, HarmCategory, HarmBlockThreshold
from tqdm import tqdm
import json
import time

PROJECT_ID = "halogen-sol-348516"
LOCATION = "us-central1"
INPUT_FILE = "train_dataset_clean.jsonl"
OUTPUT_FILE = "train_dataset_rewritten.jsonl"

vertexai.init(project=PROJECT_ID, location=LOCATION)

active_model = None
for name in ["gemini-2.0-flash-001", "gemini-1.5-flash"]:
    try:
        m = GenerativeModel(name)
        m.generate_content("test")
        active_model = m
        print(f"Connected to: {name}")
        break
    except:
        continue

if not active_model:
    raise RuntimeError("Could not connect to any Gemini model.")

# Load input data
data = []
with open(INPUT_FILE, 'r') as f:
    for line in f:
        data.append(json.loads(line))
print(f"Loaded {len(data)} samples")

# Resume logic
processed_hashes = set()
if os.path.exists(OUTPUT_FILE):
    with open(OUTPUT_FILE, 'r') as f:
        for line in f:
            try:
                entry = json.loads(line)
                processed_hashes.add(hash(entry.get('input', '')[:500]))
            except:
                pass
    print(f"Resuming: {len(processed_hashes)} already done")
````

````python
# ===== CELL 5: Run Rewriting Loop (PARALLEL — 20 workers) =====
# ~15-30 min instead of 5 hrs
from concurrent.futures import ThreadPoolExecutor, as_completed
import threading

safety = {HarmCategory.HARM_CATEGORY_DANGEROUS_CONTENT: HarmBlockThreshold.BLOCK_ONLY_HIGH}

REWRITE_PROMPT = """You are a Senior Software Engineer writing a code review for a pull request.

Given a code diff and an existing review comment, rewrite the review to be:
1. SPECIFIC - Reference actual code, variable names, or line changes from the diff
2. ACTIONABLE - Tell the developer exactly what to change and why
3. CRITICAL - If there's a bug, security issue, or bad practice, call it out clearly
4. CONCISE - 1-3 sentences max

If the diff has a security issue (hardcoded secrets, SQL injection, etc.), start with "SECURITY:" and explain the risk.
If the diff has a bug (missing null checks, wrong logic, etc.), start with "BUG:" and explain what will break.
If the diff is just a style/naming issue, explain why the naming matters.

If the diff is trivial (whitespace, formatting only) and the review adds no value, output: {{"keep": false}}

Diff:
{diff}

Existing review:
{comment}

Output JSON only: {{"keep": true/false, "rewritten": "your improved review"}}"""

stats = {"total": 0, "rewritten": 0, "filtered": 0, "skipped": 0, "errors": 0}
file_lock = threading.Lock()
stats_lock = threading.Lock()

# Filter out already-processed samples
samples_to_process = []
for sample in data:
    diff_hash = hash(sample['input'][:500])
    if diff_hash in processed_hashes:
        with stats_lock:
            stats["skipped"] += 1
    else:
        samples_to_process.append(sample)

print(f"Skipped {stats['skipped']} already processed. Processing {len(samples_to_process)} remaining...")

output_file = open(OUTPUT_FILE, "a")

def process_sample(sample):
    diff_hash = hash(sample['input'][:500])

    for attempt in range(5):
        try:
            response = active_model.generate_content(
                REWRITE_PROMPT.format(
                    diff=sample['input'][:5000],
                    comment=sample['output'],
                ),
                generation_config={"response_mime_type": "application/json"},
                safety_settings=safety,
            )

            try:
                raw = response.text.replace("```json", "").replace("```", "").strip()
                result = json.loads(raw)
            except json.JSONDecodeError:
                with stats_lock:
                    stats["errors"] += 1
                return

            keep_val = result.get("keep")
            if keep_val in [True, "true", "True"]:
                rewritten = result.get("rewritten", "").strip()
                if rewritten and len(rewritten) >= 20:
                    entry = {
                        "instruction": "Review this code change.",
                        "input": sample['input'],
                        "output": rewritten,
                        "repo": sample.get('repo', ''),
                    }
                    with file_lock:
                        output_file.write(json.dumps(entry) + "\n")
                        output_file.flush()
                        processed_hashes.add(diff_hash)
                    with stats_lock:
                        stats["rewritten"] += 1
            else:
                with stats_lock:
                    stats["filtered"] += 1

            return  # Success

        except Exception as e:
            if "429" in str(e) or "ResourceExhausted" in str(e):
                time.sleep(20 * (attempt + 1))
                continue
            else:
                with stats_lock:
                    stats["errors"] += 1
                    if stats["errors"] <= 5:
                        print(f"\nError: {str(e)[:100]}")
                return

# Run parallel
MAX_WORKERS = 20
with ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:
    futures = {executor.submit(process_sample, s): s for s in samples_to_process}
    for future in tqdm(as_completed(futures), total=len(futures), unit="sample"):
        future.result()
        with stats_lock:
            stats["total"] += 1
        if stats["total"] % 1000 == 0:
            print(f"\nProgress: {stats}")

output_file.close()
print(f"\nDone! {stats}")
print(f"Output: {OUTPUT_FILE}")
````

````python
# ===== CELL 6: Load Model for SFT =====
from unsloth import FastLanguageModel
import torch

max_seq_length = 8192

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="unsloth/Qwen2.5-Coder-7B-Instruct",
    max_seq_length=max_seq_length,
    dtype=None,
    load_in_4bit=False,
)

# FIX: pad_token != eos_token (Qwen2.5 bug)
tokenizer.pad_token = "<|vision_pad|>"

model = FastLanguageModel.get_peft_model(
    model,
    r=32,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj",
                    "gate_proj", "up_proj", "down_proj"],
    lora_alpha=32,
    lora_dropout=0,
    bias="none",
    use_gradient_checkpointing="unsloth",
    random_state=42,
)

print("Model loaded")
````

````python
# ===== CELL 7: Load & Format Rewritten Dataset =====
import json
from datasets import Dataset

data = []
with open('train_dataset_rewritten.jsonl', 'r') as f:
    for line in f:
        data.append(json.loads(line))

print(f"Loaded {len(data)} samples")

SYSTEM_MSG = "You are a Senior Software Engineer reviewing code changes. Provide clear, actionable feedback."

def format_prompt(sample):
    """Use tokenizer.apply_chat_template — NO manual ChatML."""
    messages = [
        {"role": "system", "content": SYSTEM_MSG},
        {"role": "user", "content": f"Review this code diff:\n\n```diff\n{sample['input'][:6000]}\n```"},
        {"role": "assistant", "content": sample['output']},
    ]
    text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=False)
    return {"text": text}

dataset = Dataset.from_list(data)
dataset = dataset.map(format_prompt)
dataset = dataset.train_test_split(test_size=0.05, seed=42)
print(f"Train: {len(dataset['train'])}, Val: {len(dataset['test'])}")

# Verify: check the formatted text looks correct
print("\n--- Sample (first 500 chars) ---")
print(dataset['train'][0]['text'][:500])
````

````python
# ===== CELL 8: SFT Training =====
from trl import SFTTrainer
from transformers import TrainingArguments

trainer = SFTTrainer(
    model=model,
    tokenizer=tokenizer,
    train_dataset=dataset["train"],
    eval_dataset=dataset["test"],
    dataset_text_field="text",
    max_seq_length=max_seq_length,
    dataset_num_proc=4,
    packing=True,
    args=TrainingArguments(
        per_device_train_batch_size=4,
        gradient_accumulation_steps=4,
        warmup_steps=50,
        num_train_epochs=3,
        learning_rate=5e-5,
        fp16=False,
        bf16=True,
        logging_steps=25,
        optim="adamw_8bit",
        weight_decay=0.01,
        lr_scheduler_type="cosine",
        seed=42,
        output_dir="outputs-sft-v2",
        eval_strategy="steps",
        eval_steps=100,
        save_strategy="steps",
        save_steps=200,
    ),
)

print("Starting SFT training...")
trainer_stats = trainer.train()
print("SFT training complete")
````

````python
# ===== CELL 9: Test SFT Model =====
FastLanguageModel.for_inference(model)

test_diffs = [
    ("SECURITY", """@@ -12,7 +12,7 @@ def connect(self):\n-        password = os.environ.get('DB_PASS')\n+        password = \"admin123\"\n         return db.connect(password)"""),
    ("BUG", """@@ -5,7 +5,7 @@ def get_user(user_id):\n-        user = db.query(User).filter(User.id == user_id).first()\n+        user = db.query(User).filter(User.id == user_id)\n         return user.name"""),
    ("STYLE", """@@ -1,5 +1,5 @@\n-def calculateTotalPrice(items):\n+def calc(i):\n     total = 0\n-    for item in items:\n+    for x in i:\n         total += x.price"""),
]

for i, (label, diff) in enumerate(test_diffs, 1):
    messages = [
        {"role": "system", "content": SYSTEM_MSG},
        {"role": "user", "content": f"Review this code diff:\n\n```diff\n{diff}\n```"},
    ]
    inputs = tokenizer.apply_chat_template(
        messages, tokenize=True, add_generation_prompt=True, return_tensors="pt"
    ).to("cuda")

    with torch.no_grad():
        outputs = model.generate(
            input_ids=inputs, max_new_tokens=256, temperature=0.3,
            top_p=0.9, pad_token_id=tokenizer.pad_token_id,
        )

    response = tokenizer.decode(outputs[0], skip_special_tokens=True)
    review = response.split("assistant")[-1].strip()
    print(f"\n{'='*50}\nTEST {i} ({label})\n{'='*50}\nMODEL REVIEW: {review}")
````

````python
# ===== CELL 10: Save SFT LoRA =====
model.save_pretrained("code-reviewer-lora-v2")
tokenizer.save_pretrained("code-reviewer-lora-v2")
print("SFT LoRA saved")
````

````python
# ===== CELL 11: Merge SFT into Base (needed for DPO reference) =====
model.save_pretrained_merged("sft-merged-v2", tokenizer, save_method="merged_16bit")
print("SFT model merged into base")
````

````python
# ===== CELL 12: Generate DPO Pairs (vLLM Batch Inference) =====
# Using ALL 12k samples (was 5k) with ROUGE-L quality filter
# Consistent 6000-char truncation to match SFT training distribution
import os
os.environ["VLLM_WORKER_MULTIPROC_METHOD"] = "spawn"
os.environ["VLLM_USE_V1"] = "0"

import json, gc
import torch
from rouge_score import rouge_scorer

# Free Unsloth model memory
try:
    del model, trainer
except:
    pass
torch.cuda.empty_cache()
gc.collect()
print("GPU memory freed")

from transformers import AutoTokenizer
from vllm import LLM, SamplingParams

tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-Coder-7B-Instruct")
tokenizer.pad_token = "<|vision_pad|>"

SYSTEM_MSG = "You are a Senior Software Engineer reviewing code changes. Provide clear, actionable feedback."
rl_scorer = rouge_scorer.RougeScorer(['rougeL'], use_stemmer=True)

# Load ALL rewritten data (not just 5k)
rewritten_data = []
with open('train_dataset_rewritten.jsonl', 'r') as f:
    for line in f:
        rewritten_data.append(json.loads(line))
print(f"Loaded {len(rewritten_data)} samples for DPO pair generation")

# Build prompts — truncate to 6000 chars (matches SFT training distribution)
prompts = []
for sample in rewritten_data:
    diff = sample['input'][:6000]
    messages = [
        {"role": "system", "content": SYSTEM_MSG},
        {"role": "user", "content": f"Review this code diff:\n\n```diff\n{diff}\n```"},
    ]
    text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    prompts.append(text)

print(f"Built {len(prompts)} prompts")

# Load merged SFT model with vLLM
print("Loading model with vLLM...")
llm = LLM(
    model="sft-merged-v2",
    tokenizer="Qwen/Qwen2.5-Coder-7B-Instruct",
    max_model_len=4096,
    dtype="auto",
    enforce_eager=True,
    gpu_memory_utilization=0.85,
)
sampling_params = SamplingParams(temperature=0.7, top_p=0.9, max_tokens=256)

print("Running batch inference...")
outputs = llm.generate(prompts, sampling_params)
print(f"Generated {len(outputs)} responses")

# Build DPO pairs with ROUGE-L quality filter
dpo_pairs = []
skipped_short = 0
skipped_similar = 0
skipped_dissimilar = 0

for i, (sample, output) in enumerate(zip(rewritten_data, outputs)):
    chosen_review  = sample['output']
    rejected_review = output.outputs[0].text.strip()

    # Skip too-short generations (likely truncation artifacts)
    if len(rejected_review) < 20:
        skipped_short += 1
        continue

    # ROUGE-L similarity filter:
    #   > 0.85: too similar — no useful preference signal
    #   < 0.05: too different — likely garbage / hallucinated response
    rl = rl_scorer.score(chosen_review, rejected_review)['rougeL'].fmeasure
    if rl > 0.85:
        skipped_similar += 1
        continue
    if rl < 0.05:
        skipped_dissimilar += 1
        continue

    diff = sample['input'][:6000]
    pair = {
        "prompt": [
            {"role": "system", "content": SYSTEM_MSG},
            {"role": "user", "content": f"Review this code diff:\n\n```diff\n{diff}\n```"},
        ],
        "chosen":   [{"role": "assistant", "content": chosen_review}],
        "rejected": [{"role": "assistant", "content": rejected_review}],
    }
    dpo_pairs.append(pair)

print(f"\nGenerated {len(dpo_pairs)} DPO pairs")
print(f"  Skipped (too short):      {skipped_short}")
print(f"  Skipped (too similar RL>0.85): {skipped_similar}")
print(f"  Skipped (too different RL<0.05): {skipped_dissimilar}")

with open('dpo_pairs.jsonl', 'w') as f:
    for pair in dpo_pairs:
        f.write(json.dumps(pair) + '\n')
print("Saved to dpo_pairs.jsonl")

del llm
torch.cuda.empty_cache()
gc.collect()
print("vLLM memory freed — ready for DPO training")
````

````python
# ===== CELL 13: Load Merged SFT + New LoRA for DPO =====
# Fresh load required. The merged SFT base acts as the reference model.
# DPOTrainer disables the new LoRA adapter to compute reference logprobs.

from unsloth import FastLanguageModel
import torch

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="sft-merged-v2",
    max_seq_length=4096,
    dtype=None,
    load_in_4bit=False,
)

tokenizer.pad_token = "<|vision_pad|>"

# Smaller LoRA for DPO (less aggressive than SFT)
model = FastLanguageModel.get_peft_model(
    model,
    r=16,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj",
                    "gate_proj", "up_proj", "down_proj"],
    lora_alpha=16,
    lora_dropout=0,
    bias="none",
    use_gradient_checkpointing="unsloth",
    random_state=42,
)

SYSTEM_MSG = "You are a Senior Software Engineer reviewing code changes. Provide clear, actionable feedback."
print("Merged SFT model loaded with new LoRA for DPO")
````

````python
# ===== CELL 14: Load DPO Dataset =====
import json
from datasets import Dataset

dpo_data = []
with open('dpo_pairs.jsonl', 'r') as f:
    for line in f:
        dpo_data.append(json.loads(line))

print(f"Loaded {len(dpo_data)} DPO pairs")

dataset = Dataset.from_list(dpo_data)
dataset = dataset.train_test_split(test_size=0.05, seed=42)
print(f"Train: {len(dataset['train'])}, Val: {len(dataset['test'])}")

# Verify conversational format
s = dataset['train'][0]
print(f"\nKeys: {list(s.keys())}")
print(f"prompt[0]: {s['prompt'][0]}")
print(f"chosen[0] (first 100 chars): {str(s['chosen'][0])[:100]}")
print(f"rejected[0] (first 100 chars): {str(s['rejected'][0])[:100]}")
````

````python
# ===== CELL 15: DPO Training (Fixed — anti-reward-hacking settings) =====
# Key changes vs previous run that reward-hacked:
#   beta 0.1 → 0.05   (tighter KL constraint, less room to deviate from reference)
#   lr 5e-6 → 2e-6    (more conservative — previous run saturated at step 150)
#   2 epochs → 1 + max_steps=500 cap (val loss plateaued at step ~400 before)
#   early stopping (patience=3 eval steps = 150 training steps)
from unsloth import PatchDPOTrainer
PatchDPOTrainer()

from trl import DPOTrainer, DPOConfig
from transformers import EarlyStoppingCallback

dpo_trainer = DPOTrainer(
    model=model,
    ref_model=None,  # Merged SFT base = reference (LoRA disable trick)
    args=DPOConfig(
        beta=0.05,                    # was 0.1 — tighter KL penalty
        learning_rate=2e-6,           # was 5e-6 — more conservative
        per_device_train_batch_size=2,
        gradient_accumulation_steps=4,
        num_train_epochs=1,           # was 2 — val loss converged by step ~400
        max_steps=500,                # hard cap — safety net
        warmup_ratio=0.1,
        bf16=True,
        logging_steps=10,
        optim="adamw_8bit",
        seed=42,
        output_dir="outputs-dpo-v3",
        eval_strategy="steps",
        eval_steps=50,
        save_strategy="steps",
        save_steps=100,
        load_best_model_at_end=True,  # needed for early stopping
        metric_for_best_model="eval_loss",
        greater_is_better=False,
        remove_unused_columns=False,
        max_length=2048,
        max_prompt_length=1500,
    ),
    train_dataset=dataset["train"],
    eval_dataset=dataset["test"],
    processing_class=tokenizer,
    callbacks=[EarlyStoppingCallback(early_stopping_patience=3)],
)

print("Starting DPO training (beta=0.05, lr=2e-6, max_steps=500, early stopping)...")
dpo_trainer.train()
print("DPO training complete")
````

````python
# ===== CELL 16: Test DPO Model =====
model.eval()

test_diffs = [
    ("SECURITY", """@@ -12,7 +12,7 @@ def connect(self):\n-        password = os.environ.get('DB_PASS')\n+        password = \"admin123\"\n         return db.connect(password)"""),
    ("BUG", """@@ -5,7 +5,7 @@ def get_user(user_id):\n-        user = db.query(User).filter(User.id == user_id).first()\n+        user = db.query(User).filter(User.id == user_id)\n         return user.name"""),
    ("STYLE", """@@ -1,5 +1,5 @@\n-def calculateTotalPrice(items):\n+def calc(i):\n     total = 0\n-    for item in items:\n+    for x in i:\n         total += x.price"""),
]

print("=" * 60)
print("QUALITATIVE TEST — DPO model on 3 diffs")
print("=" * 60)

for i, (label, diff) in enumerate(test_diffs, 1):
    messages = [
        {"role": "system", "content": SYSTEM_MSG},
        {"role": "user", "content": f"Review this code diff:\n\n```diff\n{diff}\n```"},
    ]
    inputs = tokenizer.apply_chat_template(
        messages, tokenize=True, add_generation_prompt=True, return_tensors="pt"
    ).to("cuda")

    with torch.no_grad():
        outputs = model.generate(
            input_ids=inputs, max_new_tokens=256, temperature=0.3,
            top_p=0.9, pad_token_id=tokenizer.pad_token_id,
        )

    response = tokenizer.decode(outputs[0], skip_special_tokens=True)
    review = response.split("assistant")[-1].strip()
    print(f"\n{'='*50}\nTEST {i} ({label})\n{'='*50}")
    print(f"DIFF:\n{diff}")
    print(f"\nDPO REVIEW:\n{review}")

# ── Systematic eval on validation set ──
print("\n" + "=" * 60)
print("SYSTEMATIC EVAL — DPO vs SFT on validation set")
print("=" * 60)

from rouge_score import rouge_scorer as rs_module
import numpy as np

eval_scorer = rs_module.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)

def quick_eval(model_obj, tok, samples, max_samples=50):
    """Generate predictions on a sample of the val set and compute ROUGE + reward."""
    preds, refs, diffs_eval = [], [], []
    for sample in samples[:max_samples]:
        diff = sample['input'][:6000]
        msgs = [
            {"role": "system", "content": SYSTEM_MSG},
            {"role": "user", "content": f"Review this code diff:\n\n```diff\n{diff}\n```"},
        ]
        inp = tok.apply_chat_template(msgs, tokenize=True, add_generation_prompt=True, return_tensors="pt").to("cuda")
        with torch.no_grad():
            out = model_obj.generate(input_ids=inp, max_new_tokens=256, do_sample=False, pad_token_id=tok.pad_token_id)
        text = tok.decode(out[0], skip_special_tokens=True).split("assistant")[-1].strip()
        preds.append(text)
        refs.append(sample['output'])
        diffs_eval.append(diff)
    r1, r2, rl = [], [], []
    for p, ref in zip(preds, refs):
        scores = eval_scorer.score(ref, p)
        r1.append(scores['rouge1'].fmeasure)
        r2.append(scores['rouge2'].fmeasure)
        rl.append(scores['rougeL'].fmeasure)
    return {"rouge1": np.mean(r1), "rouge2": np.mean(r2), "rougeL": np.mean(rl)}, preds, diffs_eval

# Load the raw val data for eval
val_raw = []
with open('train_dataset_rewritten.jsonl') as f:
    all_data = [json.loads(l) for l in f]
val_raw = all_data[-100:]  # last 100 as a quick eval set

print("Running DPO model on 50 val samples...")
dpo_metrics, dpo_val_preds, dpo_val_diffs = quick_eval(model, tokenizer, val_raw, max_samples=50)

print(f"\nDPO Model (val):")
for k, v in dpo_metrics.items():
    print(f"  {k}: {v:.4f}")

print("\n✓ If DPO metrics are better than or equal to SFT, set UPSTREAM_MODEL='dpo' in dr.grpo.ipynb Cell 6")
print("✗ If DPO metrics are worse, keep UPSTREAM_MODEL='sft' (reward hacking not fully fixed)")
````
